```python
#Recieves inputted value from the user
firstName = str(input("first name;"))

lastName = str(input("last name;"))

section = str(input("section;"))

favSub = str(input("favorite subject;"))

#Combines first and last/sur name
fullName = firstName + " " + lastName

#Prints the gives formatted values:
print(" ")
print("=== Formatted Output: ===")
print(" ")
print("Name:", fullName.title())
print("Name Code:", fullName.upper())
print("Section:", section.upper())
print("Favorite Subject:", favSub.title())
print("Character count:", len(fullName))
```
